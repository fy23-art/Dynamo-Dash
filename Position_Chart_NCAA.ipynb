```python
import matplotlib.font_manager as fm

fontPathBold = r"C:\Users\freya\Downloads\Dash\Fonts\Fonts\DDCHardware-Regular.otf"
fontPathNBold = r"C:\Users\freya\Downloads\Dash\Fonts\Fonts\DDCHardware-Condensed.otf"

font_prop = fm.FontProperties(fname=fontPathBold, size=16)
font_prop_legend = fm.FontProperties(fname=fontPathNBold, size=12)

def plot_player_positions(data, player_name, team_name):
    df = data[
        (data['player_name'] == player_name) & 
        (data['team_name'] == team_name) & 
        (data['event_type_name'] == 'Block') &
        (data['event_type_name'] == 'Pressure') &
        (data['event_type_name'] == 'Tackle') &
        (data['event_type_name'] == 'Clearance') &
        (data['event_type_name'] == 'Foul Commited') &
        (data['event_type_name'] == 'Duel') & 
        (data['event_type_name'] != 'Substitution')
    ]
    grouped = df.groupby('player_position_name').agg({
        'location_x': 'mean',
        'location_y': 'mean',
        'player_position_name': 'count'
    }).rename(columns={'player_position_name': 'count'}).reset_index()

    grouped = grouped[grouped['count'] > 50]


    total_events = grouped['count'].sum()
    grouped['percent'] = grouped['count'] / total_events * 100

    grouped['scaled_x'] = grouped['location_x'] * (105 / 120)
    grouped['scaled_y'] = 68 - (grouped['location_y'] * (68 / 80))

    import matplotlib.colors as mcolors
    orange_cmap = mcolors.LinearSegmentedColormap.from_list("orange_cmap", ["#ffffff", "#FF6900"])
    norm = plt.Normalize(grouped['percent'].min(), grouped['percent'].max())

    fig, ax = plt.subplots(figsize=(12, 8))
    horizfull_pitch(pitch='#f4f4f4', line='#222', ax=ax)

    for _, row in grouped.iterrows():
        initials = ''.join([w[0] for w in row['player_position_name'].split()])
        label_text = initials
        color = orange_cmap(norm(row['percent']))
        ax.scatter(row['scaled_x'], row['scaled_y'], s=row['count']*20, alpha=0.7, 
                   label=initials, color=color, edgecolors='black')
        ax.text(row['scaled_x'], row['scaled_y'], label_text, ha='center', va='center', 
                fontsize=14, color='black', fontproperties=font_prop)

    plt.title(f"{player_name} ({team_name}) - Average Position Locations", fontproperties=font_prop)

    legend_rows = grouped.sort_values('percent', ascending=False)
    handles = []
    labels = []
    for _, row in legend_rows.iterrows():
        initials = ''.join([w[0] for w in row['player_position_name'].split()])
        percent = f"{row['percent']:.1f}%"
        label = f"{initials} ({percent})"
        color = orange_cmap(norm(row['percent']))  
        handles.append(plt.Line2D([0], [0], marker='o', color='w', label=label,
                                  markerfacecolor=color, markersize=10))
        labels.append(label)
    ax.legend(handles, labels, title="Position", bbox_to_anchor=(1.05, 1), 
              loc='upper left', fontsize=9, prop=font_prop_legend, 
              title_fontproperties=font_prop_legend)
    plt.tight_layout()

    filename = f"{player_name.replace(' ', '_')}_{team_name.replace(' ', '_')}_{season}_positions.png"
    plt.savefig(rf"C:\Users\freya\Downloads\Dash\{filename}", bbox_inches='tight')
    plt.show()

players = [
    ("Alana Cook", "Kansas City"),
    ("Ellis Wheeler", "Kansas City"),
    ("Kayla Sharples", "Kansas City"),
    ("Elizabeth Ball", "Kansas City")
]

fig, axes = plt.subplots(2, 2, figsize=(18, 12))
axes = axes.flatten()

for ax, (player_name, team_name) in zip(axes, players):
    df = data[
        (data['player_name'] == player_name) & 
        (data['team_name'] == team_name) & 
        (data['event_type_name'] != 'Substitution')
    ]
    grouped = df.groupby('player_position_name').agg({
        'location_x': 'mean',
        'location_y': 'mean',
        'player_position_name': 'count'
    }).rename(columns={'player_position_name': 'count'}).reset_index()
    grouped = grouped[grouped['count'] > 50]
    total_events = grouped['count'].sum()
    grouped['percent'] = grouped['count'] / total_events * 100
    grouped['scaled_x'] = grouped['location_x'] * (105 / 120)
    grouped['scaled_y'] = 68 - (grouped['location_y'] * (68 / 80))
    import matplotlib.colors as mcolors
    orange_cmap = mcolors.LinearSegmentedColormap.from_list("orange_cmap", ["#ffffff", "#FF6900"])
    norm = plt.Normalize(grouped['percent'].min(), grouped['percent'].max())
    horizfull_pitch(pitch='#f4f4f4', line='#222', ax=ax)
    for _, row in grouped.iterrows():
        initials = ''.join([w[0] for w in row['player_position_name'].split()])
        label_text = initials
        color = orange_cmap(norm(row['percent']))
        ax.scatter(row['scaled_x'], row['scaled_y'], s=row['count']*20, alpha=0.7, 
                   label=initials, color=color, edgecolors='black')
        ax.text(row['scaled_x'], row['scaled_y'], label_text, ha='center', va='center', 
                fontsize=14, color='black', fontproperties=font_prop)
    ax.set_title(f"{player_name} ({team_name})", fontproperties=font_prop)
    ax.axis('off')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

fontPathBold = r"C:\Users\freya\Downloads\Dash\Fonts\Fonts\DDCHardware-Regular.otf"
font_prop = fm.FontProperties(fname=fontPathBold, size=12)

def plot_horizontal_defensive_line(data, players_list, team_name):
    # Filter for defensive events and specific players
    defensive_events = ['Block', 'Pressure', 'Tackle', 'Clearance', 'Foul Committed', 'Duel']
    
    df = data[
        (data['team_name'] == team_name) & 
        (data['player_name'].isin([p[0] for p in players_list])) &
        (data['event_type_name'].isin(defensive_events))
    ]

    # Group by player to find average coordinates
    summary = df.groupby(['player_name', 'player_position_name']).agg({
        'location_x': 'mean',
        'location_y': 'mean'
    }).reset_index()

    # Scaling to a standard 105x68 pitch
    summary['scaled_x'] = summary['location_x'] * (105 / 120)
    summary['scaled_y'] = summary['location_y'] * (68 / 80)

    fig, ax = plt.subplots(figsize=(14, 10))
    
    horizfull_pitch(pitch='#ffffff', line='#222', ax=ax) 

    summary = summary.sort_values('scaled_y')

    ax.plot(summary['scaled_x'], summary['scaled_y'], color='black', linewidth=3, zorder=1)

    #Plot the Circles and Labels
    for i, row in summary.iterrows():
        pos_name = row['player_position_name'].lower()
        circle_color = '#99C18F' if 'right' in pos_name else '#FDE74C'
        
        ax.scatter(row['scaled_x'], row['scaled_y'], s=1500, color=circle_color, 
                   edgecolors='#333', linewidth=1.5, zorder=2)
        
        initials = ''.join([w[0] for w in row['player_position_name'].split()]).upper()
        
        ax.text(row['scaled_x'], row['scaled_y'], initials, ha='center', va='center', 
                fontsize=10, fontproperties=font_prop, zorder=3)

    # Defensive Height Line
    avg_height = summary['scaled_x'].mean()
    ax.axvline(avg_height, color='grey', linestyle='--', alpha=0.6, zorder=0)
    ax.text(avg_height + 1, 5, f"{avg_height:.2f}m", color='grey', rotation=90, fontproperties=font_prop)

    plt.title(f"{team_name} - Defensive Average Height", fontproperties=font_prop, pad=20)
    plt.tight_layout()
    plt.show()

players_to_plot = [
    ("Alana Cook", "Kansas City"),
    ("Ellis Wheeler", "Kansas City"),
    ("Kayla Sharples", "Kansas City"),
    ("Elizabeth Ball", "Kansas City")
]

plot_horizontal_defensive_line(data, players_to_plot, "Kansas City")
```
